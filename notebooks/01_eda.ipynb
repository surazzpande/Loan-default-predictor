```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../data/credit_risk_dataset.csv")
df.head()
```

```python
df.info()                 # column types
df.isnull().sum()         # missing values
df.describe()             # stats, look for weird values (e.g. age = 144)
df["loan_status"].value_counts(normalize=True)  # how imbalanced? (~22% default)
```

```python
sns.countplot(data=df, x="loan_grade", hue="loan_status")
plt.title("Default by Loan Grade"); plt.show()

sns.boxplot(data=df, x="loan_status", y="loan_percent_income")
plt.show()
```
